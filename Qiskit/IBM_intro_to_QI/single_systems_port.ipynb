```python
from IPython.core.display_functions import display
from qiskit import __version__
import numpy as np
from qiskit.visualization import array_to_latex
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_histogram
from qiskit.quantum_info import Operator
from qiskit import QuantumCircuit
```

```python
"""Vector Math"""
ket0 = np.array([[1], [0]])
ket1 = np.array([[0], [1]])
print(ket0 / 2 + ket1 / 2)

"""Matrix Math"""
M1 = np.array([[1,1], [0,0]])
M2 = np.array([[1,0],[0,1]])
M = M1 / 2 + M2 / 2
print(M)

"""Matrix Multiplication"""
print(f"M1@M2 = {M1@M2}")
```

```python
u = Statevector([1 / np.sqrt(2), 1 / np.sqrt(2)])
v = Statevector([(1+2.0j) / 3, -2/3])
w = Statevector([1/3, 2/3])

display(u.draw("text"))
display(u.draw("latex_source"))
display(u.draw("latex"))
"""check for valid quantum state vector (Euclidian norm = 1)"""
display(u.is_valid())
display(w.is_valid())
```

Display and Measure V

```python
display(v.draw("latex"))

#takes a probabalistic measurement of the quantum state
outcome, state = v.measure()
print(f"Measured: {outcome}\nPost-measurement state:")
display(state.draw("latex"))

mult_meas = v.sample_counts(1000)
plot_histogram(mult_meas)
```

```python
Y = Operator([[0, -1j], [1j, 0]])
H = Operator([[1/np.sqrt(2), 1/np.sqrt(2)], [1/np.sqrt(2), -1/np.sqrt(2)]])

display(Y.draw("latex"))
display(H.draw("latex"))
```

```python
#Simple state vector
v = Statevector([1,0])
print("v:")
display(v.draw("latex"))

#evolve applies the opperations
v = v.evolve(Y)
print("Y:")
display(Y.draw("latex"))
print("vY:")
display(v.draw("latex"))

v = v.evolve(H)
print("H:")
display(H.draw("latex"))
print("vYH")
display(v.draw("latex"))
```

Quantum Circuits:

```python
circuit = QuantumCircuit(1)

#predefined quantum circuit operators
circuit.h(0)
circuit.t(0)
circuit.h(0)
circuit.s(0)
circuit.y(0)

display(circuit.draw(output="mpl"))
```
